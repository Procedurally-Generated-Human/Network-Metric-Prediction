```python
import algorithms, helpers
import torch
import torch.nn as nn
```

```python
df = helpers.load_puffer_dataset()
qs = [0.1, 0.25, 0.75, 0.9]
horizons=[1,2,4,8,16]
n_lags=10
sample_size=800000

model_basic = algorithms.LSTM_regressor(in_d=6, hidden_d=128, out_d=1, num_layers=4, dropout=0.2)
basic_model_param = {
    'model': model_basic,
    'batch_size': 512,
    'epochs': 100,
    'objective': nn.HuberLoss(),
    'optim': torch.optim.AdamW(model_basic.parameters(), lr=1e-4),
    'device': torch.device("cuda" if torch.cuda.is_available() else "cpu")
}

model_quantile = algorithms.LSTM_quantile(in_d=6, hidden_d=128, out_d=len(qs), num_layers=2, dropout=0.2)
quantile_model_param = {
    'model': model_quantile,
    'batch_size': 512,
    'epochs': 100,
    'objective': helpers.quantile_huber_loss,
    'optim': torch.optim.AdamW(model_quantile.parameters(), lr=1e-4),
    'device': torch.device("cuda" if torch.cuda.is_available() else "cpu")
}
```

```python
for metric in ['rtt', 'delivery_rate']:
    algorithms.run_lstm_comparison_basic(df, basic_model_param, target_metric=metric, n_lags=n_lags, sample_size=sample_size, horizons=horizons)
```

```python
for metric in ['rtt', 'delivery_rate']:
    algorithms.run_lstm_comparison_quantile(df, quantile_model_param, target_metrics=metric, n_lags=n_lags, sample_size=sample_size, horizons=horizons, qs=qs)
```
